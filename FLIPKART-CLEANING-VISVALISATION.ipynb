```python
#Products overpriced but low sales

import numpy as np

price_threshold = df['final_price'].quantile(0.75)
sales_threshold = df['units_sold'].quantile(0.25)

overpriced = df[(df['final_price'] > price_threshold) & (df['units_sold'] < sales_threshold)]
overpriced[['product_name', 'final_price', 'units_sold']].head()

```

```python
#High rating product but low sales

high_rating = df['rating'].quantile(0.75)
low_sales = df['units_sold'].quantile(0.25)

hidden_gems = df[(df['rating'] > high_rating) & (df['units_sold'] < low_sales)]
hidden_gems[['product_name', 'rating', 'units_sold']].head()
```

```python
#Segment product by price

conditions = [
    df['final_price'] < df['final_price'].quantile(0.33),
    df['final_price'] < df['final_price'].quantile(0.66)
]

choices = ['Low', 'Medium']

df['price_segment'] = np.select(conditions, choices, default='High')

df['price_segment'].value_counts()
```

```python
# Risky products

risky = df[(df['rating'] < 3) & (df['return_policy_days'] > 10)]
risky[['product_name', 'rating', 'return_policy_days']].head()
```
